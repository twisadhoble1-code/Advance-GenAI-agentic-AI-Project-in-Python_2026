Variables.py

```python
first_name = "Twisa"
middle_name = "Mahendra"
last_name = "Dhoble"
country = "India"
age = 25
is_married = False
skills = ["Python", "Java", "C++"]
person_info = {
    "first_name": "Raffealla",
    "middle_name": "Justine",
    "last_name": "Petrino",
    "Country": "South Africa",
    "City": "Cape of Good Hope"
}

# Printing the values of the stored variables

print("First Name:", first_name)
print("Middle Name:", middle_name)
print("Last Name:", last_name)
print("Country:", country)
print("Age:", age)
print("Is Married:", is_married)
print("Skills:", skills)
print("Person Info:", person_info)

# Declaring multiple variables in a single line

first_name, middle_name, last_name, country, age, is_married = "Twisa", "Mahendra", "Dhoble", "India", 25, False

```
